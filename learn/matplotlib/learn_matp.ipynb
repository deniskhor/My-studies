pie charts (круговые диаграммы)

```python
import matplotlib.pyplot as plt
import numpy as np

categories = ['Freshmen', 'Sophohmores', 'Juniors', 'Seniors']
values = np.array([300, 250, 275, 225])

plt.pie(values, labels=categories, 
        autopct='%1f') #%1.1f, %1.1f%%

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Названия категорий, которые будут подписаны на круговой диаграмме
categories = ['Freshmen', 'Sophohmores', 'Juniors', 'Seniors']

# Значения для каждой категории
# По этим числам Matplotlib считает долю каждого сектора
values = np.array([300, 250, 275, 225])

# Цвет каждого сектора по порядку:
# Freshmen -> red, Sophohmores -> yellow, Juniors -> blue, Seniors -> green
colors = ['red', 'yellow', 'blue', 'green']

plt.pie(
    values,                 # данные для круговой диаграммы
    labels=categories,      # подписи секторов
    autopct='%1f',          # показывает процент внутри сектора
    colors=colors,          # цвета секторов
    explode=[0, 0, 0, 0.1], # "выдвигает" сектор Seniors наружу на 0.1
    shadow=True,            # добавляет тень под диаграммой
    startangle=90           # поворачивает диаграмму, старт с угла 90 градусов
)

# Заголовок графика
plt.title('Bro Code College')

# Показывает график на экране
plt.show()
```

scatter graphs (графики рассеяния)

```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8]) # hours studied
y1 = np.array([55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]) # grades

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8]) # hours studied
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97]) # grades

plt.scatter(x1, y1, color='skyblue', 
            alpha = 0.9, # от 0.1 до 1.0
            s = 200, 
            label = 'Class A')

plt.scatter(x2, y2, color='red', 
            alpha = 0.9, # от 0.1 до 1.0
            s = 200, 
            label = 'Class B')

plt.title('Test Scores')
plt.xlabel('Hours Studied')
plt.ylabel('Grade')

plt.legend()
plt.show()
```

histograms (гистограмы)

```python
import matplotlib.pyplot as plt
import numpy as np

scores = np.random.normal(loc=80, scale=10, size=100)
scores = np.clip(scores, 0, 100)

plt.hist(scores, bins = 20,
        color = 'lightgreen', 
        edgecolor = 'black')

plt.title('Exam Scores')
plt.xlabel('Score')
plt.ylabel('# of students')

plt.show()
```

subplots (“подграфики” или “несколько графиков в одной фигуре”)

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4,5])

figure, axes = plt.subplots(2, 2)

axes[0, 0].plot(x, x*2, color = 'red')
axes[0, 0].set_title('x*2')

axes[0, 1].bar(x, x**2, color = 'blue')
axes[0, 1].set_title('x**2')

axes[1, 0].plot(x, x**3, color = 'green')
axes[1, 0].set_title('x**3')

axes[1, 1].plot(x, x**4, color = 'purple')
axes[1, 1].set_title('x**4')

plt.tight_layout()
plt.show()
```

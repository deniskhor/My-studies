```python
# Статический метод в Python — это функция внутри класса, которая не использует данные объекта (self) или класса (cls). 
# Он создается с помощью декоратора @staticmethod и вызывается без создания копии класса.

# @staticmethod — это метод, который не привязан ни к какому объекту. Его можно вызывать напрямую от класса, без создания экземпляра.


class Employee:
    
    def __init__(self, name, position):
        self.name = name
        self.position = position
        
    def get_info(self):
        return f'{self.name} = {self.position}'
    
    @staticmethod
    def is_valid_position(position):
        valid_positions = ['Manager', 'Cashier', 'Cook', 'Janitor']
        return position in valid_positions
    
    
employee1 = Employee('Eugune', 'Manager')
employee2 = Employee('Squidward', 'Cashier')
employee3 = Employee('Spongebob', 'Cook')
    
print(Employee.is_valid_position('Cook'))
print(Employee.is_valid_position('Rocket Scientist'))


print(employee1.get_info())
print(employee2.get_info())
print(employee3.get_info())
```

```python
# Задание 1.2: Атрибуты класса vs атрибуты объекта
# Цель: Научиться различать общие характеристики (атрибуты класса) и уникальные (атрибуты объекта), а также понять, как они взаимодействуют.

# Условие
# Ты создаешь систему учета студентов в школе.

# Создай класс Student со следующими атрибутами класса:
# school = "Школа №7" (название школы)
# grade = 10 (класс по умолчанию для всех новых студентов)

# Создай трех студентов (экземпляры класса):
# student1 — имя "Анна", возраст 16 лет
# student2 — имя "Максим", возраст 17 лет
# student3 — имя "Елена", возраст 15 лет

# Добавь каждому студенту индивидуальные атрибуты:
# Для student1: предмет, который он любит — "Математика", оценка за четверть — 5
# Для student2: предмет, который он любит — "Физика", оценка за четверть — 4
# Для student3: предмет, который он любит — "Литература", оценка за четверть — 5

# Выведи на экран следующую информацию:
# Имя и возраст каждого студента (через обращение к атрибутам объекта)
# Школу для каждого студента (через обращение к экземпляру)
# Класс для каждого студента (через обращение к экземпляру)
# Любимый предмет и оценку для каждого студента

# Поэкспериментируй с изменением атрибутов:
# Измени атрибут класса school на "Гимназия №3"
# Выведи школу для всех трех студентов (через экземпляры)
# У student2 измени атрибут grade на 11 (индивидуально)
# Выведи класс для student1, student2 и student3
# Создай нового студента student4 (имя и возраст придумай сам) и выведи его школу и класс — проверь, какие значения у него будут

# Исследуй внутреннее устройство:
# Выведи __dict__ каждого студента
# Выведи __dict__ класса Student
# Сравни, что лежит в __dict__ у студентов и у класса

class Student:
    school = "Школа №7"
    grade = 10
    
student1 = Student()
student1.name = "Анна"
student1.age = 16
student1.subject = "Математика"
student1.rating = 5

student2 = Student()
student2.name = "Максим"
student2.age = 17
student2.subject = "Физика"
student2.rating = 4

student3 = Student()
student3.name = "Елена"
student3.age = 15
student3.subject = "Литература"
student3.rating = 5

# 1. Вывод информации 
print(f'''Имя первого студента: {student1.name}, второго: {student2.name}, третьего: {student3.name}.
Обучаются все они в учебном заведении {student1.school}.
Все обучаются в {student1.grade} классе.
Любимый предмет у первого студента: {student1.subject}, оценка за четверть: {student1.rating}; у второго: {student2.subject}, оценка: {student2.rating}; у третьего: {student3.subject}, оценка: {student3.rating}
''')

print('---------------------------')

# 2. Меняем школу (атрибут класса)
Student.school = "Гимназия №3"
print(f'Школа student1: {student1.school}')
print(f'Школа student2: {student2.school}')
print(f'Школа student3: {student3.school}')

print('---------------------------')

# 3. Меняем класс у student2 (создаем атрибут объекта)
student2.grade = 11
print(f'Класс student1: {student1.grade}, student2: {student2.grade}, student3: {student3.grade}')

print('---------------------------')

# 4. Создаем student4
student4 = Student()
student4.name = "Денис"
student4.age = 18
student4.subject = "Информатика"
student4.rating = 5

print(f'student4 обучается в {student4.school}, в {student4.grade} классе.')

print('---------------------------')

# 5. Исследуем внутреннее устройство
print(f'__dict__ student1: {student1.__dict__}')
print(f'__dict__ student2: {student2.__dict__}')
print(f'__dict__ student3: {student3.__dict__}')
print(f'__dict__ student4: {student4.__dict__}')
print(f'__dict__ класса Student: {Student.__dict__}')




```

```python
class Book:
    library = "Городская библиотека" 
    cover = "Твердый" 

# Создаем объекты и добавляем атрибуты
book1 = Book()
book1.name = "Война и мир"
book1.author = "Толстой"
book1.year = 1869

book2 = Book()
book2.name = "Преступление и наказание"
book2.author = "Достоевский"
book2.year = 1866

book3 = Book()
book3.name = "Мастер и Маргарита"
book3.author = "Булгаков"
book3.year = 1967

# Используем setattr() для добавления страниц (ИСПРАВЛЕНО!)
setattr(book1, 'pages', 1225)
setattr(book2, 'pages', 672)
setattr(book3, 'pages', 480)

# Выводим информацию с помощью getattr()
print("=== Информация о книгах ===")
print(f"Название book1: {getattr(book1, 'name')}")
print(f"Название book2: {getattr(book2, 'name')}")
print(f"Название book3: {getattr(book3, 'name')}")

print(f"Автор book1: {getattr(book1, 'author')}")
print(f"Автор book2: {getattr(book2, 'author')}")
print(f"Автор book3: {getattr(book3, 'author')}")

print(f"Страниц book1: {getattr(book1, 'pages')}")
print(f"Страниц book2: {getattr(book2, 'pages')}")
print(f"Страниц book3: {getattr(book3, 'pages')}")

# Рейтинг с значением по умолчанию (ИСПРАВЛЕНО!)
print(f"Рейтинг book1: {getattr(book1, 'rating', 'Нет рейтинга')}")

print("\n=== Проверка наличия атрибутов ===")
print(f"У book1 есть author? {hasattr(book1, 'author')}")  # True
print(f"У book1 есть rating? {hasattr(book1, 'rating')}")  # False
print(f"У book2 есть pages? {hasattr(book2, 'pages')}")    # True
print(f"У book3 есть cover? {hasattr(book3, 'cover')}")    # True (из класса)

# Удаляем атрибуты
delattr(book1, 'pages')
delattr(book2, 'author')
delattr(book3, 'year')

print("\n=== После удаления атрибутов ===")
print(f"У book1 есть pages? {hasattr(book1, 'pages')}")    # False
print(f"У book2 есть author? {hasattr(book2, 'author')}")  # False
print(f"У book3 есть year? {hasattr(book3, 'year')}")      # False

# Добавляем статусы
setattr(book1, 'status', 'В наличии')
setattr(book2, 'status', 'Выдана')
setattr(book3, 'status', 'В ремонте')

print("\n=== Статусы книг ===")
print(f"Статус book1: {getattr(book1, 'status')}")
print(f"Статус book2: {getattr(book2, 'status')}")
print(f"Статус book3: {getattr(book3, 'status')}")

# Исследуем внутреннее устройство
print("\n=== Внутреннее устройство ===")
print(f"__dict__ book1: {book1.__dict__}")
print(f"__dict__ book2: {book2.__dict__}")
print(f"__dict__ book3: {book3.__dict__}")
print(f"__dict__ класса Book: {Book.__dict__}")  # ИСПРАВЛЕНО!

# Объяснение
print("\n=== Объяснение ===")
print("Почему в __dict__ книг нет атрибута library?")
print("Потому что library — это атрибут класса. Он хранится в __dict__ класса Book")
print("и является общим для всех экземпляров. В __dict__ каждого объекта")
print("хранятся только те атрибуты, которые были добавлены конкретному объекту.")
```

SERIES

```python
# Series - одномерный массив данных с метками (индексами)

import pandas as pd # импорт библиотеки

data = [100, 102, 104] # исходный список

series = pd.Series(data) # создание Series из списка

print(series) # вывод: индекс слева, значение справа, dtype внизу
```

Все наши данные расположены в одном столбце. У нас появился одномерный массив данных - Series. В левой стороне находятся метки, которые начинаются с нуля. Внизу приведены метаданные об этой серии. Тип данных, которые мы храним - 64-битные целые числа

```python
# Series - автоматическое определение типа данных

import pandas as pd

data = [100.1, 102.2, 104.3] # вещественные числа

series = pd.Series(data) # pandas сам определяет тип float64

print(series)
```

При изменении типа данных поменялись и метаданные. Теперь это 64-битные вещественные числа.

```python
# Series - работа со строками

import pandas as pd

data = ['A', 'B', 'C'] # строковые данные

series = pd.Series(data) # тип object (так pandas хранит строки)

print(series)
```

```python
# Series - булевы значения

import pandas as pd

data = [True, False, False] # логический тип

series = pd.Series(data) # тип bool

print(series)
```

```python
# Series - собственный индекс

import pandas as pd 

data = [100, 102, 104] 

series = pd.Series(data, index =['apartment #1', 'apartment #2', 'apartment #3']) # index - задаём свои метки

print(series) # теперь индексы - строки, а не числа
```

```python
# Series - доступ по метке (label-based)

import pandas as pd 

data = [100, 102, 104] 

series = pd.Series(data, index =['apartment #1', 'apartment #2', 'apartment #3']) 

print(series.loc['apartment #1']) # loc[] - доступ по имени индекса
```

```python
# Series - изменение значения

import pandas as pd 

data = [100, 102, 104] 

series = pd.Series(data, index =['apartment #1', 'apartment #2', 'apartment #3']) 

series.loc['apartment #1'] = 200 # присваиваем новое значение по метке

print(series) # значение изменилось
```

Прежние данные:

apartment #1    100
apartment #2    102
apartment #3    104
dtype: int64

```python
# Series - доступ по позиции (position-based)

import pandas as pd 

data = [100, 102, 104] 

series = pd.Series(data, index =['apartment #1', 'apartment #2', 'apartment #3']) 

print(series.iloc[0]) # iloc[] - доступ по целочисленной позиции (независимо от названий индексов)
```

iloc[0] всегда работает с целочисленной позицией элемента независимо от того, какие названия у индексов. В данном случае элемент 'apartment #1' находится на позиции 0, а так как iloc[0] обращается именно к нулевой позиции, то они совпадают и ошибки не возникает — iloc просто берет первый по порядку элемент, игнорируя его строковую метку.

Таким образом, мы можем использовать доступ по целочисленному местоположению или доступ по метке.

```python
# Series - фильтрация

import pandas as pd 

data = [100, 102, 104, 200, 202]

series = pd.Series(data, index =['a', 'b', 'c', 'd', 'e'])

print(series[series >= 200]) # series >= 200 создаёт маску (True/False), затем применяет её
```

```python
# Series - фильтрация с другим условием

import pandas as pd 

data = [100, 102, 104, 200, 202]

series = pd.Series(data, index =['a', 'b', 'c', 'd', 'e'])

print(series[series < 200]) # оставляем только элементы меньше 200
```

FILTERING

```python
# Filtering - отбор строк по условию

import pandas as pd

df = pd.read_csv('pokemon.csv')

tall_pokemon = df[df['Height'] >= 2] # df[условие] - создаёт маску и применяет её

print(tall_pokemon)
```

tall_pokemon = df[df['Height'] >= 2] — это фильтрация:
df['Height'] — берём столбец «Height» (рост)
df['Height'] >= 2 — проверяем, у каких покемонов рост больше или равен 2 
df[...] — оставляем в таблице только те строки, где условие выполняется
Всё это сохраняем в переменную tall_pokemon (высокие покемоны)

отфильтрованные_данные = df[df['Название_столбца'] >= число]

«Дай мне df, но только те строки, где в столбце Height значение больше или равно 2»
Внешние квадратные скобки df[...] означают «выбери строки», а внутри — условие отбора.

```python
# Filtering - ИЛИ (|)

import pandas as pd

df = pd.read_csv('pokemon.csv')

water_pokemon = df[(df['Type1'] == 'Water') | # оператор | (ИЛИ)
                    (df['Type2'] == 'Water')] # условия в скобках для приоритета

print(water_pokemon)
```

```python
# Filtering - И (&)

import pandas as pd

df = pd.read_csv('pokemon.csv')

ff_pokemon = df[(df['Type1'] == 'Fire') & # оператор & (И) - оба условия
                (df['Type2'] == 'Flying')]

print(ff_pokemon)
```

APPLY

```python
# Apply - применение пользовательских функций

import pandas as pd

df = pd.read_csv('pokemon.csv')

# 1. Применение лямбда-функции

df['Height_cm'] = df['Height'].apply(lambda x: x * 100) # перевод метров в сантиметры

print(df[['Name', 'Height', 'Height_cm']].head(10))
```

```python
# Apply - применение пользовательской функции

import pandas as pd

df = pd.read_csv('pokemon.csv')

# 2. Создание категорий веса

def categorize_weight(weight):
    if weight < 50:
        return 'Light'
    elif weight < 200:
        return 'Medium'
    else:
        return 'Heavy'

df['Weight_Category'] = df['Weight'].apply(categorize_weight) # применяем функцию к колонке

print(df[['Name', 'Weight', 'Weight_Category']].head(10))
```

```python
# Apply - работа со строками (axis=1)

import pandas as pd

df = pd.read_csv('pokemon.csv')

# 3. Применение функции к нескольким колонкам

def calculate_bmi(row):
    return row['Weight'] / (row['Height'] ** 2) # ИМТ = вес / рост²

df['BMI'] = df.apply(calculate_bmi, axis=1) # axis=1 - применяем к каждой строке

print(df[['Name', 'Height', 'Weight', 'BMI']].head(10))
```

PIVOT TABLE

```python
# Pivot Table - сводные таблицы (альтернатива groupby)

import pandas as pd

df = pd.read_csv('pokemon.csv')

# 1. Простая сводная таблица

pivot = pd.pivot_table(df, 
                       values='Height', # значения для агрегации
                       index='Type1', # строки (группировка)
                       aggfunc='mean') # функция агрегации

print(pivot) # средний рост по типам
```

```python
# Pivot Table - множественная агрегация

import pandas as pd

df = pd.read_csv('pokemon.csv')

# 2. Несколько колонок и функций

pivot = pd.pivot_table(df,
                       values=['Height', 'Weight'], # несколько колонок
                       index='Type1', # строки
                       aggfunc={'Height': ['mean', 'max'], # разные функции для колонок
                                'Weight': ['mean', 'max']})

print(pivot) # средние и максимальные значения для каждого типа
```

```python
# Pivot Table - двумерная группировка

import pandas as pd

df = pd.read_csv('pokemon.csv')

# 3. Строки и колонки (как в Excel)

pivot = pd.pivot_table(df,
                       values='Height',
                       index='Type1', # строки
                       columns='Legendary', # колонки (0 или 1)
                       aggfunc='mean',
                       fill_value=0) # fill_value - чем заполнять пропуски

print(pivot) # средний рост легендарных и обычных покемонов по типам
```

MERGING

```python
# Merging - объединение таблиц

import pandas as pd

# 1. Создание двух таблиц для примера

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Bulbasaur', 'Ivysaur', 'Venusaur', 'Charmander'],
    'Type': ['Grass', 'Grass', 'Grass', 'Fire']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Height': [0.7, 1.0, 2.0, 0.6],
    'Weight': [6.9, 13.0, 100.0, 8.5]
})

print('Таблица 1:')
print(df1)
print('\nТаблица 2:')
print(df2)

# 2. Объединение по общему ключу

merged = pd.merge(df1, df2, on='ID') # объединяем по колонке ID

print('\nОбъединённая таблица:')
print(merged)
```

```python
# Merging - разные типы объединений

import pandas as pd

# 1. Таблицы с разными ID

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5],
    'Name': ['Bulbasaur', 'Ivysaur', 'Venusaur', 'Charmander', 'Charmeleon']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3, 6, 7],
    'Height': [0.7, 1.0, 2.0, 0.6, 1.1]
})

print('Таблица 1:')
print(df1)
print('\nТаблица 2:')
print(df2)

# 2. Inner join (по умолчанию)

inner = pd.merge(df1, df2, on='ID', how='inner') # только совпадающие ID
print('\nInner join (только совпадающие):')
print(inner)

# 3. Left join

left = pd.merge(df1, df2, on='ID', how='left') # все строки из df1
print('\nLeft join (все из df1):')
print(left)

# 4. Right join

right = pd.merge(df1, df2, on='ID', how='right') # все строки из df2
print('\nRight join (все из df2):')
print(right)

# 5. Outer join

outer = pd.merge(df1, df2, on='ID', how='outer') # все строки из обеих
print('\nOuter join (все из обеих):')
print(outer)
```
